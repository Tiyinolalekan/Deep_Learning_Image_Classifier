```python
import pandas as pd
import shutil
from pathlib import Path


base = "/root/.cache/kagglehub/datasets/paramaggarwal/fashion-product-images-small/versions/1"
csv_path = base + "/styles.csv"
images_dir = base + "/images"
output_dir = "/content/sorted"
images_dir = Path(images_dir)
output_dir = Path(output_dir)

# Using on_bad_lines='skip' to bypass the malformed lines in the CSV
df = pd.read_csv(csv_path, on_bad_lines='skip')
df.head()
print(df)
```

***SPLITTING DATA INTO TRAIN TEST AND VALIDATION***

```python

from pathlib import Path
from PIL import Image
import random


RAW_DIR = Path(
    r"/content/sorted"
)

OUTPUT_DIR = Path("/content/clothes")

SOURCES = {
    "dress": [
        RAW_DIR / "dresses",
    ],
    "hat": [
        RAW_DIR / "hat",
        RAW_DIR / "caps",
    ],
    "hoodie": [
        RAW_DIR / "sweatshirts",
        RAW_DIR / "sweaters",
    ],
    "outwear": [
        RAW_DIR / "jackets",
        RAW_DIR / "blazers",
        RAW_DIR / "rain jacket",
        RAW_DIR / "shrug",
        RAW_DIR / "waistcoat",
        RAW_DIR / "nehru jackets",
    ],
    "t-shirt": [
        RAW_DIR / "tshirts",
        RAW_DIR / "tops",
        RAW_DIR / "lounge tshirts",
        RAW_DIR / "camisoles",
    ],
    "pants": [
        RAW_DIR / "jeans",
        RAW_DIR / "track pants",
        RAW_DIR / "trousers",
        RAW_DIR / "capris",
        RAW_DIR / "rain trousers",
        RAW_DIR / "jeggings",
        RAW_DIR / "leggings",
        RAW_DIR / "lounge pants",
        RAW_DIR / "tights",
    ],
    "shoes": [
        RAW_DIR / "casual shoes",
        RAW_DIR / "sports shoes",
        RAW_DIR / "formal shoes",
    ],
    "shirt": [
        RAW_DIR / "shirts",
    ],
    "skirt": [
        RAW_DIR / "skirts",
    ],
    "shorts": [
        RAW_DIR / "shorts",
        RAW_DIR / "lounge shorts",
    ],
}
IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png", ".webp", ".bmp"
}

random.seed(42)


def find_images(source_folders):
    images = []

    for folder in source_folders:
        if not folder.exists():
            print(f"Missing folder: {folder}")
            continue

        for path in folder.rglob("*"):
            if path.suffix.lower() not in IMAGE_EXTENSIONS:
                continue

            # Exclude unreadable or non-image files
            try:
                with Image.open(path) as img:
                    img.verify()
                images.append(path)
            except Exception:
                continue

    return images


# Collect images from all sources
class_images = {}

for label, folders in SOURCES.items():
    class_images[label] = find_images(folders)
    print(f"{label}: {len(class_images[label])} images found")

# Do not proceed with missing classes
missing = [
    label for label, images in class_images.items()
    if not images
]

if missing:
    raise ValueError(
        f"No images found for {missing}. "
        "Check folder paths and add the missing dataset."
    )

# Balance all classes to the size of the smallest class
#target_count = min(
   # len(images) for images in class_images.values()
#)

#print(f"\nUsing {target_count} images per class.")

splits = {
    "train": 0.8,
    "val": 0.1,
    "test": 0.1,
}

# Start with a clean output folder
if OUTPUT_DIR.exists():
    shutil.rmtree(OUTPUT_DIR)

for label, images in class_images.items():
    random.shuffle(images)
    selected = images
    end = len(selected)

    train_end = int(end * 0.8)
    val_end = int(end * 0.9)

    partitions = {
        "train": selected[:train_end],
        "val": selected[train_end:val_end],
        "test": selected[val_end:],
    }

    for split, files in partitions.items():
        destination = OUTPUT_DIR / split / label
        destination.mkdir(parents=True, exist_ok=True)

        for index, source in enumerate(files):
            # Unique names prevent collisions between datasets
            filename = f"{index:06d}_{source.name}"
            shutil.copy2(source, destination / filename)

        print(f"{split}/{label}: {len(files)} images")

print(f"\nDataset created at: {OUTPUT_DIR}")
```
